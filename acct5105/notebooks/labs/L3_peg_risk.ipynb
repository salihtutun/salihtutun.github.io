Try USDT-USD for two years. If `yfinance` is blocked, **do not fake

```python
live_ok = False
used = "class freezer (live download skipped or failed)"
# Smoke tests set COURSE_SMOKE=1 so we do not hit the network.
want_live = os.environ.get("COURSE_SMOKE") != "1"

if want_live:
    try:
        import yfinance as yf

        df = yf.download("USDT-USD", period="2y", progress=False, auto_adjust=True)
        px = df["Close"].dropna()
        if hasattr(px, "columns"):  # yfinance sometimes returns a 1-col frame
            px = px.iloc[:, 0]
        if len(px) < 30:
            raise RuntimeError("too few rows")
        gap = (px - 1.0).abs()
        worst = float(gap.max())
        when = str(gap.idxmax())[:10]
        fig, ax = plt.subplots(figsize=(7.4, 3.8))
        ax.plot(px.index, px.values, color="#2c4a6e", lw=1)
        ax.axhline(1.0, color="#c45c26", ls="--", lw=1, label="poster $1")
        ax.set_title(f"USDT-USD · worst |price−1| = {worst:.4f} on {when}")
        ax.set_ylabel("USD per USDT")
        ax.legend(frameon=False)
        fig.savefig(OUT / "L3-depeg.png", dpi=140, bbox_inches="tight")
        plt.show()
        live_ok = True
        used = f"USDT-USD via yfinance; worst gap {worst:.4f} on {when}"
        print(used)
    except Exception as exc:
        print("LIVE DOWNLOAD FAILED - do not invent prices.")
        print("reason:", type(exc).__name__, exc)
        print("PDF: either retry later, or screenshot Naranjo's depeg figure.")

if not live_ok:
    # Class picture only — labeled so nobody thinks this is USDT.
    cards, scoops = 10_000, 200
    fig, ax = plt.subplots(figsize=(7.4, 3.8))
    ax.axhline(1.0, color="#c45c26", ls="--", label="poster: 1 card = 1 scoop")
    ax.scatter([0], [scoops / cards], color="#2c4a6e", s=60, zorder=3)
    ax.set_xlim(-0.5, 3)
    ax.set_ylim(0, 1.15)
    ax.set_xticks([])
    ax.set_ylabel("scoops per card")
    ax.set_title("L3-depeg: CLASS freezer - not live market data")
    ax.legend(frameon=False)
    fig.savefig(OUT / "L3-depeg.png", dpi=140, bbox_inches="tight")
    plt.show()
    print("saved class-freezer L3-depeg.png")
    print("If you needed a real coin: Naranjo Stablecoins and Peg Risk.")
print("saved", OUT / "L3-depeg.png")

```

## Saturday rush (feeds Q4–Q5)

10,000 gift cards, 200 scoops. The poster still says 1 card = 1 scoop.



```python
scoops = 200
cards = 10_000
print("if everyone ran tonight, each card is really", scoops / cards, "scoops")


def rush(shown, freezer):
    honored = min(shown, freezer)
    freezer = freezer - honored
    return honored, freezer


h1, f1 = rush(150, 200)
print("wave 1: 150 show up -> honored", h1, "freezer", f1)
h2, f2 = rush(80, f1)
print("wave 2: 80 show up  -> honored", h2, "freezer", f2)
print("The rest of wave 2 hold a PDF.")

```
