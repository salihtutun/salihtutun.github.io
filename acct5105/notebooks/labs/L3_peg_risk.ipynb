## Saturday rush (feeds Q4–Q5)

10,000 gift cards, 200 scoops. The poster still says 1 card = 1 scoop.



```python
scoops = 200  # coins leaving the bowl in stress test
cards = 10_000  # set cards
print("if everyone ran tonight, each card is really", scoops / cards, "scoops")  # print for the PDF or screen


def rush(shown, freezer):  # define helper rush
    honored = min(shown, freezer)  # set honored
    freezer = freezer - honored  # set freezer
    return honored, freezer  # hand back the result


h1, f1 = rush(150, 200)  # set h1
print("wave 1: 150 show up -> honored", h1, "freezer", f1)  # print for the PDF or screen
h2, f2 = rush(80, f1)  # set h2
print("wave 2: 80 show up  -> honored", h2, "freezer", f2)  # print for the PDF or screen
print("The rest of wave 2 hold a PDF.")  # print for the PDF or screen

```
