## Words we use today

Read this once. If a later cell uses the word, come back here.

| Word | In this class it means |
|---|---|
| **Constant product** | x × y = k. After a trade the product stays k. |
| **Spot** | y / x — CLASSUSD per one extra TEAM, right now. |
| **Average** | What you actually paid for this chunk (paid ÷ TEAM bought). |
| **Slippage** | You moved the bowl, so you paid worse than the starting spot. |
| **IL snapshot** | Jar vs just holding the two stacks, valued at the new spot. Not a fee. |



```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np

OUT = Path("lab_output")
OUT.mkdir(exist_ok=True)

X0, Y0 = 200.0, 100.0  # TEAM, CLASSUSD
K = X0 * Y0
print("start x, y, k =", X0, Y0, K)
print("start spot (CLASSUSD per TEAM) =", Y0 / X0)

```

## Q2 — Buy until 160 TEAM remain (5 pts)

A classmate takes TEAM out of the bowl. \(k\) does not change.
Fill (a)–(e) from the print: new y, CLASSUSD paid, average, new spot,
and one sentence on slippage.



```python
def buy_team(x, y, team_out):
    """Take TEAM out of a no-fee bowl. Returns new piles, paid, new spot, average."""
    k = x * y
    new_x = x - team_out
    if new_x <= 0:
        raise ValueError("cannot buy the whole bowl")
    new_y = k / new_x
    paid = new_y - y
    team_bought = team_out
    return new_x, new_y, paid, new_y / new_x, paid / team_bought


nx, ny, paid, spot, avg = buy_team(X0, Y0, 40)  # 200 → 160 remaining
print("new x (TEAM left)     ", nx)
print("(a) new y (CLASSUSD)  ", ny)
print("(b) CLASSUSD paid     ", paid)
print("    TEAM bought       ", 40)
print("(c) average price     ", avg)
print("(d) new spot          ", spot)
print("(e) slippage vs start ", avg - (Y0 / X0), "(average minus old spot)")

# Picture: slide along x*y = k.
xs = np.linspace(80, 280, 200)
fig, ax = plt.subplots(figsize=(7.2, 4.0))
ax.plot(xs, K / xs, color="#2c4a6e", label="x × y = 20,000")
ax.scatter([X0, nx], [Y0, ny], color="#c45c26", zorder=3)
ax.annotate("start 200 / 100", (X0, Y0), xytext=(X0 + 8, Y0 + 8))
ax.annotate("after buy 160 / 125", (nx, ny), xytext=(nx - 70, ny + 12))
ax.set_xlabel("TEAM in the bowl (x)")
ax.set_ylabel("CLASSUSD in the bowl (y)")
ax.set_title("L2-bowl: you slide along the curve")
ax.legend(frameon=False)
fig.savefig(OUT / "L2-bowl.png", dpi=140, bbox_inches="tight")
plt.show()
print("saved", OUT / "L2-bowl.png")

```

## Q3 — Who puts TEAM back? (4 pts)

After Q2, TEAM is *richer* in CLASSUSD terms (spot went up). An arb that
wants the **old** 0.50 price must sell TEAM *into* the bowl (add TEAM,
remove CLASSUSD) until \(x=200\) again.

The cell below does that reverse trade. **Write in the PDF:** which pile
does the bot add, what is that trade doing, four sentences. Exact size
is printed; you may cite it.



```python
def sell_team_into_pool(x, y, team_in):
    """Add TEAM, take CLASSUSD out, keep k."""
    k = x * y
    new_x = x + team_in
    new_y = k / new_x
    classusd_out = y - new_y
    return new_x, new_y, classusd_out, new_y / new_x


# Inverse of buying 40 TEAM: add 40 TEAM back.
rx, ry, out, rspot = sell_team_into_pool(nx, ny, 40)
print("after selling 40 TEAM into the bowl:")
print("  x, y, spot =", rx, ry, rspot)
print("  CLASSUSD taken out of the bowl =", out)
print("spot back to start?", abs(rspot - Y0 / X0) < 1e-9)

```

## Q4 — Jar vs backpack (4 pts)

outside the bowl, valued at the **new** spot.



```python
hold_team, hold_usd = 200.0, 100.0
pool_value = nx * spot + ny
hold_value = hold_team * spot + hold_usd
print("new spot                    ", spot)
print("withdraw the jar            ", pool_value)
print("just hold 200 TEAM + 100 USD", hold_value)
print("gap (IL snapshot)           ", hold_value - pool_value)
print("There were no fees in this problem. Do not call the gap a fee.")

```
