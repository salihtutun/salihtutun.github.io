## Failure stories (write one that actually happened)

1. **Wrong country on the envelope** — MetaMask on Ethereum, class on Sepolia.
2. **No stamps** — faucet dry, gas = 0.
3. **Typo locker** — airdrop to `0xABC` instead of a roster address.

Document tx hash + the fix. That paragraph is points, not extra credit.



```python
# ===== paste facts AFTER you deploy =====
network = "Sepolia"
chain_id = 11155111
contract = "0x_PASTE_REAL_ADDRESS"
token = "EAGE"
admin = "0x_TEAM_ADMIN"
deploy_tx = "0x_DEPLOY_HASH"
airdrop_tx = "0x_AIRDROP_HASH"
failed_tx = "0x_OR_NONE"
failed_why = "wrong network / no gas / cap / typo — pick the true one"
failed_fix = "what you did next"

classmate_balances = [
    ("Ava", "0xAAA", 200),
    ("Ben", "0xBBB", 200),
    ("Cara", "0xCCC", 200),
]


def looks_like_eth_addr(s: str) -> bool:
    return s.startswith("0x") and len(s) >= 10 and "PASTE" not in s


print("RUNBOOK")
print(f"1. Network: {network} (chain id {chain_id})")
print(f"2. Contract: {contract} ({token})")
print(f"3. Admin/minter/pauser: {admin}")
print("4. Balance: explorer token page or balanceOf(address)")
print("5. If pending forever: faucet + correct network")
print(f"Deploy tx: {deploy_tx}")
print(f"Airdrop tx: {airdrop_tx}")
print(f"Failure: {failed_tx} because {failed_why}")
print(f"Fix: {failed_fix}")
print("Three mailboxes:")
for name, addr, bal in classmate_balances:
    print(f"  {name} {addr} {bal} {token}")

if not looks_like_eth_addr(contract):
    print("\n!! contract still looks like a placeholder -- finish Remix first.")

```
