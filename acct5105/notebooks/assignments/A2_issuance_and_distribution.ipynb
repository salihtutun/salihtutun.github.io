```python
# ===== edit me =====
name = "Eagecoin"
symbol = "EAGE"
decimals = 0
cap = 10_000
minter = "team-owner-wallet"
pauser = "team-owner-wallet"
lost_key_plan = "We pause via a second teammate key / we accept mint is dead."

roster_size = 30          # count the real roster
airdrop_each = 200
pool_reserve = 2_000
treasury = 2_000
primary_method = "airdrop"  # airdrop | sale | earn
why_not_the_others = "A gift, not a sale, so we are not pitching a 10x."

planned = airdrop_each * roster_size + pool_reserve + treasury
fits = planned <= cap

print(f"{name} ({symbol})  decimals={decimals}  cap={cap}")
print(f"minter={minter}  pauser={pauser}")
print(f"lost-key plan: {lost_key_plan}")
print(f"method: {primary_method}  |  {why_not_the_others}")
print(f"airdrop {roster_size} × {airdrop_each} = {airdrop_each * roster_size}")
print(f"pool {pool_reserve} + treasury {treasury}")
print(f"planned mint {planned}  fits cap? {fits}")
if not fits:
    print("MATH FAIL — this is an automatic A2 hit. Shrink something.")
else:
    print("MATH OK — paste this table into your PDF.")

```
