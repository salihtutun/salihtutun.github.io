```python
# Market note numbers. Replace with YOUR bowl (Uniswap or class shadow book).

bowl_kind = "class notebook bowl"  # or 'testnet Uniswap'
team_in_pool = 2000  # starting TEAM reserve x
classusd_in_pool = 1000  # starting CLASSUSD reserve y
k = team_in_pool * classusd_in_pool  # invariant k = x * y
spot = classusd_in_pool / team_in_pool  # CLASSUSD per TEAM at the margin

# One external swap: stranger bought 50 TEAM. Fee-less class math:
x, y = team_in_pool, classusd_in_pool  # copy starting piles
team_out = 50  # TEAM leaving the bowl
new_x = x - team_out  # TEAM left after the sip
new_y = k / new_x  # CLASSUSD that keeps k fixed
paid = new_y - y  # CLASSUSD the stranger paid
new_spot = new_y / new_x  # new CLASSUSD per TEAM
avg_paid = paid / team_out  # average price on this gulp

print(f"Bowl: {bowl_kind}")  # say which market you used
print(f"k = {k}")  # post k on the market note
print(f"start spot {spot:.4f} CLASSUSD per TEAM")  # listing at seed
print(
    f"after stranger bought {team_out}: paid {paid:.2f}, "
    f"avg {avg_paid:.4f}, new spot {new_spot:.4f}"
)  # post-swap numbers for the PDF
print("IL: we are the LP. Gap vs hold stacks is not a fee (class bowl = no fees).")  # IL reminder
print("Overnight? Write one honest TEAM vs CLASSUSD sentence in the PDF.")  # market note closer
print("Passport: paste x, y, spot, and the stranger hash into the evidence pack.")  # Final Project feed

```
