## Worked example (do not copy as your submission)

**Community:** students in this section. Join = your wallet on the roster.

**Problem:** homework help is messy IOUs in group chat. They don't transfer.
Ava helps Ben, Ben never helps Cara, Cara wanted Ava.

**Unit:** we *think* in minutes. We still *price* against CLASSUSD so the
class has one measuring stick.

**What it is:** your asset + the team's promise of tutoring time
(1 EAGE = 5 minutes).

**Supply idea:** cap 10,000. Only the team mints. Burn when minutes are delivered.

**Week 4 use:** Ava pays Ben 6 EAGE for 30 minutes of algebra.



```python
# ===== YOUR COIN INTENT — edit every string =====
# Talk like a human. If a 15-year-old glazes over, rewrite.

team_names = ["Your Name", "Teammate"]
community = "Who is in? Who is out? How do you join?"
problem = "One sentence. A payment or coordination problem — not 'we want a blockchain'."
unit_of_account = "Do people think in CLASSUSD, in your coin, or in a real good?"
asset_and_promise = "Holder asset AND whose promise?"
supply_idea = "Who prints, what caps it, what retires it?"
week4_use = "One concrete trade a classmate will actually do."
working_title = "Eagecoin"
ticker = "EAGE"  # 3–5 letters


def too_vague(text: str) -> bool:
    bad = ["blockchain", "decentralized", "revolution", "moon", "web3"]
    return any(w in text.lower() for w in bad) and "cannot" not in text.lower()


print("=" * 60)
print(f"{working_title} ({ticker})")
print("Team:", ", ".join(team_names))
print("-" * 60)
for label, val in [
    ("Community", community),
    ("Problem", problem),
    ("Unit", unit_of_account),
    ("Asset + promise", asset_and_promise),
    ("Supply idea", supply_idea),
    ("Week 4 use", week4_use),
]:
    print(f"\n{label}:\n  {val}")
    if too_vague(val):
        print("  !! sounds like a slogan -- add a real-world verb (pay, honor, redeem, sit).")
print("=" * 60)
print("Copy the block above into a 1-page PDF. 11 pt. Names at the top.")

```
