## 1. One hoodie, two sentences

- “The hoodie is **8 CLASSUSD**.” That is the *unit* (how we think).
- “Ava paid **16 EAGE**.” That is the *rail* (what moved).

Both can be true if the implied price is 0.50 CLASSUSD per EAGE.

**Demo.** Same good. Price in CLASSUSD. Settle in EAGE.



```python
# Demo — unit vs rail. Always name both.

HOODIE_CLASSUSD = 8
EAGE_PER_CLASSUSD = 2          # 0.50 CLASSUSD per EAGE
rail_amount = HOODIE_CLASSUSD * EAGE_PER_CLASSUSD

print("UNIT  hoodie costs", HOODIE_CLASSUSD, "CLASSUSD")
print("RAIL  Ava pays", rail_amount, "EAGE")
print("implied CLASSUSD per EAGE", HOODIE_CLASSUSD / rail_amount)

```

**What you just saw.** Price in one unit, settle on one rail — write both, every time.

### Keep this straight

**If you remember one thing.** “8 CLASSUSD” is the unit. “16 EAGE” is the rail. Both can be true.

**Common mix-up.** “We paid in crypto, so we priced in crypto.” You can price in dollars and settle in punches.

**Why Eagecoin cares.** A4’s circulation log needs: who, what good, unit price, rail, tx hash. Missing the unit makes the market unreadable.
