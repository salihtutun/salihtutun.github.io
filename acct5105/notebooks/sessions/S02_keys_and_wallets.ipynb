## 1. A hash is a tamper-evident sticker

Gym bag + sticker with a 64-character code. Move a sock → the code no
longer matches. Anyone can still *read* the bag. The sticker only detects
tampering.

**Demo.** One digit changes. The fingerprint looks unrelated.



```python
# Demo — hash detects tampering. It does not hide the payment.

import hashlib


def sha256(text: str) -> str:
    return hashlib.sha256(text.encode("utf-8")).hexdigest()


original = "Ava pays Ben 2 CLASSUSD"
tweaked = "Ava pays Ben 3 CLASSUSD"  # one digit -- a tiny fraud
print("original:", sha256(original))
print("tweaked: ", sha256(tweaked))
print("same?", sha256(original) == sha256(tweaked))
print("There is no “close.” The first 8 characters should look unrelated.")

```

**What you just saw.** A hash is a fingerprint. Change one character and you get a new print — not encryption.

### Keep this straight

**If you remember one thing.** A hash detects tampering. It does not encrypt the message. Anyone can still *read* the payment.

**Common mix-up.** “Crypto means the payment is secret.” In this class the payment is usually public. The *approval* is locked.

**Why Eagecoin cares.** When we later post an EAGE transfer, the hash of the transaction is the receipt we grade — not a screenshot of a balance.

## 2. A signature is a yes, not a hidden balance

**Address (public):** locker number on the gym wall. Friends can drop CLASSUSD in.
**Private key / seed:** the only key that opens it.
**Signature:** you turning the key in front of witnesses.

**Demo.** Ava's key works. Ben's key does not. A changed amount does not.



```python
# Demo — toy lock. Real wallets do not work this way.

import hmac


def toy_sign(secret: str, message: str) -> str:
    return hmac.new(secret.encode(), message.encode(), hashlib.sha256).hexdigest()


def toy_verify(secret: str, message: str, signature: str) -> bool:
    return hmac.compare_digest(toy_sign(secret, message), signature)


ava_secret = "ava-never-put-this-in-a-pdf"
ben_secret = "ben-also-fake"
payment = "send 2 CLASSUSD from Ava to Ben"
sig = toy_sign(ava_secret, payment)

print("Ava's key works?", toy_verify(ava_secret, payment, sig))
print("Ben forged it?", toy_verify(ben_secret, payment, sig))
print("Changed amount still valid?", toy_verify(ava_secret, "send 200 CLASSUSD from Ava to Ben", sig))

```

**What you just saw.** A signature is authorization (“I approved this”), not a hidden balance.

### Keep this straight

**If you remember one thing.** Ava’s key works. Ben’s key does not. A changed amount does not.

**Common mix-up.** “The signature hides how much pizza is in the fridge.” It only decides who may open the door.

**Why Eagecoin cares.** Later, only the minter key may print EAGE. That is the same idea as Ava’s toy key.
