## 1. Erase a cycle of 10

A owes B 10, B owes C 10, C owes A 10.
Ship three payments = three rails. Erase 10 from each leg = **all zeros**.

Do not call your ERC-20 a CBDC. That word is taken.

**Demo.** Perfect triangle dies.



```python
# Demo -- rows owe columns. Erase a 3-cycle.

names = ["A", "B", "C"]
owes = [
    [0, 10, 0],
    [0, 0, 10],
    [10, 0, 0],
]


def net_cycle(matrix, amount):
    m = [row[:] for row in matrix]
    m[0][1] = m[0][1] - amount
    m[1][2] = m[1][2] - amount
    m[2][0] = m[2][0] - amount
    return m


print("before", owes)
print("after 10", net_cycle(owes, 10))
print("CLASSUSD is a class chip. A CBDC is a central-bank petal. Not the same.")

```

**What you just saw.** A cycle of equal IOUs can be erased. Uneven leftovers stay. Your ERC-20 is not a CBDC.

### Keep this straight

**If you remember one thing.** A cycle of equal IOUs can be erased. Uneven leftovers stay.

**Common mix-up.** Calling EAGE or CLASSUSD a CBDC. That word is taken by a central bank.

**Why Eagecoin cares.** If three teams owe each other help, net first, then settle the leftover in CLASSUSD. Fewer txs, same economics.
