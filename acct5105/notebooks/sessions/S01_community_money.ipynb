## 2. A ledger is just a list of who has what

A **bank app** is a list. **Bitcoin** is a list with extra referees.

Ava has 10 punches. Pizza costs 2.
1. Check: 10 ≥ 2? Yes.
2. Ava: 10 − 2 = 8.
3. Cafeteria: 0 + 2 = 2.

If you forget step 3, punches vanish. That is a burn, not a payment.

**Demo.** Watch both sides move.



```python
# Demo — a payment is TWO updates.

punches = {"Ava": 10, "Ben": 10, "Cara": 10, "Dev": 2, "cafeteria": 0}
PIZZA_PRICE = 2


def buy_pizza(buyer: str) -> None:
    if punches[buyer] < PIZZA_PRICE:
        print(buyer, "cannot buy -- only", punches[buyer], "left")
        return
    punches[buyer] = punches[buyer] - PIZZA_PRICE
    punches["cafeteria"] = punches["cafeteria"] + PIZZA_PRICE
    print(buyer, "bought pizza. Balances:", punches)


buy_pizza("Ava")   # should work: Ava 8, cafeteria 2

```

**What you just saw.** A payment is two updates. Forget the cafeteria and punches vanish. Watch cafeteria gain 2.

### Keep this straight

**If you remember one thing.** Ava spent 2 and the cafeteria gained 2. That is a payment, not a burn.

**Common mix-up.** “Dev has 2 so he cannot buy.” In the Demo, 2 = pizza price, so Dev *can* buy until YOUR TURN sets him to 1.

**Why Eagecoin cares.** Every EAGE transfer later is still two updates: sender down, receiver (or cafeteria) up.

## 3. Hoarding kills the lunch line

In the 1970s a D.C. babysitting club printed *too little* paper scrip.
Families hoarded the last slip. Nobody sat. Kids and parents were still
there. The *medium of exchange* hid.

**Velocity** = punches spent this week / punches that exist.
Near 0 → you built a collectible.

**Demo.** People refuse to spend their last punch. Watch the line die.



```python
# Demo — hoarding. People will not spend their last punch.

pockets = {"Ava": 3, "Ben": 1, "Cara": 1, "Dev": 1}
stock = sum(pockets.values())  # 6 punches exist
spent = 0
PIZZA = 2


def try_buy(name: str) -> None:
    global spent
    if pockets[name] <= 1:
        print(name, "hoards the last punch -- no sale")
        return
    if pockets[name] < PIZZA:
        print(name, "cannot buy -- only", pockets[name])
        return
    pockets[name] -= PIZZA
    spent += PIZZA
    print(name, "bought. pockets", pockets)


for name in ["Ava", "Ben", "Cara", "Dev"]:
    try_buy(name)

print("spent", spent, "stock", stock, "velocity", round(spent / stock, 2))
print("Pizza still exists. The *punches* hid.")

```

**What you just saw.** Velocity near 0 means the club looks like it has money but does not *use* it.

### Keep this straight

**If you remember one thing.** Pizza still exists. The *punches* hid. That is a money problem.

**Common mix-up.** “We printed too little so people should hoard.” Hoarding is what *kills* the line.

**Why Eagecoin cares.** If classmates will not spend EAGE, you built a sticker, not lunch money. A1 must name a use, not a collection.

## 4. Asset for you, promise for the group

When **you** hold a punch, it feels like treasure (an **asset**).
When the **cafeteria** took that punch, it promised pizza (a **liability**).
Both are true.

**Demo.** 1 EAGE = 5 minutes of tutoring the team still owes.



```python
# Demo — T-account: holder treasure AND team IOU.

holders = {"Ava": 6, "Ben": 0, "Cara": 12}
MINUTES_PER_COIN = 5
team_minutes_owed = sum(holders.values()) * MINUTES_PER_COIN

print("Holder view (assets):", holders)
print("Team view (liability):", team_minutes_owed, "minutes still owed")

```

**What you just saw.** A coin is an asset for the holder and a promise for the issuer. Both lines must exist.

### Keep this straight

**If you remember one thing.** A punch is treasure to Ava and a pizza-promise for the cafeteria.

**Common mix-up.** “The token is valuable because it is on a blockchain.” Value is the promise you can still keep.

**Why Eagecoin cares.** A1 must name the community, the broken job, and the promise (minutes, cookie, invoice).
