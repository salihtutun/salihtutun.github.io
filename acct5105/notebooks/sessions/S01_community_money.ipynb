## 2. A ledger is a list. A payment is two updates

Clipboard: Ava 10, Ben 10, cafeteria 0. Pizza costs 2.

1. Check Ava ≥ 2.
2. Ava: 10 − 2 = 8.
3. Cafeteria: 0 + 2 = 2.

Forget step 3 and punches vanish. That is a **burn**, not a payment.

Three verbs: **transfer** (one down, one up), **mint** (a new balance appears), **burn** (a balance disappears). A payment is a transfer. Sender down *and* receiver up — conservation.

The two lines do not change. The **referee** does: bank, Venmo company, or many computers (Bitcoin). Mix-up: “Bitcoin is not a list.” It is a list *plus* extra referees.

**Demo.** Watch both sides move.


```python
# Demo — a payment is TWO updates.

punches = {"Ava": 10, "Ben": 10, "Cara": 10, "Dev": 2, "cafeteria": 0}
PIZZA_PRICE = 2


def buy_pizza(buyer: str) -> None:
    if punches[buyer] < PIZZA_PRICE:
        print(buyer, "cannot buy -- only", punches[buyer], "left")
        return
    punches[buyer] = punches[buyer] - PIZZA_PRICE
    punches["cafeteria"] = punches["cafeteria"] + PIZZA_PRICE
    print(buyer, "bought pizza. Balances:", punches)


buy_pizza("Ava")   # should work: Ava 8, cafeteria 2
```

### Keep this straight

**If you remember one thing.** Ava spent 2 and the cafeteria gained 2. That is a payment, not a burn.

**Common mix-up.** “Dev has 2 so he cannot buy.” In the Demo, 2 = pizza price, so Dev *can* buy until YOUR TURN sets him to 1.

## 3. Hoarding kills the lunch line (velocity)

**Velocity** = punches spent this period ÷ punches that exist.

- 3 / 30 = 0.10 → looks rich, acts frozen.
- 45 / 30 = 1.5 → some punches moved more than once. Job 1 is working.

**Babysitting co-op (1970s D.C.).** Too little paper. Families hoarded the last slip. Nobody sat. Parents and kids were still there. Shortage of the *medium of exchange*, not of children (Sweeney and Sweeney 1977; Krugman 1998).

Too *much* scrip is the other break: people dump it, the store-of-value job dies. Tonight’s Demo is the shortage.

**Sarafu** (Kenya) is a modern community money. Same test: does it circulate, or sit?

**Demo.** People refuse to spend their last punch.


```python
# Demo — hoarding. People will not spend their last punch.

pockets = {"Ava": 3, "Ben": 1, "Cara": 1, "Dev": 1}
stock = sum(pockets.values())  # 6 punches exist
spent = 0
PIZZA = 2


def try_buy(name: str) -> None:
    global spent
    if pockets[name] <= 1:
        print(name, "hoards the last punch -- no sale")
        return
    if pockets[name] < PIZZA:
        print(name, "cannot buy -- only", pockets[name])
        return
    pockets[name] -= PIZZA
    spent += PIZZA
    print(name, "bought. pockets", pockets)


for name in ["Ava", "Ben", "Cara", "Dev"]:
    try_buy(name)

print("spent", spent, "stock", stock, "velocity", round(spent / stock, 2))
print("Pizza still exists. The *punches* hid.")
```

**If you remember one thing.** Pizza still exists. The *punches* hid. That is a money problem.

**Common mix-up.** “We printed too little so people should hoard.” Hoarding is what *kills* the line.

## 4. Asset for you, promise for the group

The same punch is treasure in your pocket (**asset**) and a pizza-IOU on the cafeteria’s books (**liability**). Both lines must exist.

If the store dies, the gift card is a pretty PDF. Value is the promise you can still keep — not the fact that a row exists on a list.

An assurer asks: which two lines moved (occurrence / completeness)? Does the holder still control it (rights)? Who owes the real thing (obligations)? Is it circulating (valuation)?

**Demo.** 1 coin = 5 minutes of tutoring the team still owes.


```python
# Demo — T-account: holder treasure AND team IOU.

holders = {"Ava": 6, "Ben": 0, "Cara": 12}
MINUTES_PER_COIN = 5
team_minutes_owed = sum(holders.values()) * MINUTES_PER_COIN

print("Holder view (assets):", holders)
print("Team view (liability):", team_minutes_owed, "minutes still owed")
```
