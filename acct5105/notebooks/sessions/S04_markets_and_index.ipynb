## 1. Beta is a dance partner, not a seatbelt

**Run with me.** Do not skip ahead to YOUR TURN yet.

**Real life.** A risk desk asks: “When the S&P playlist moves, how hard did
*this* name move with it?” That habit number is **beta**. It is not a promise
that tomorrow will behave.

An **index** is a playlist. **Beta** of 2: when the playlist jumps 1%, this
coin *tends* to jump ~2% in *this* tiny sample.

Handmade numbers. No wifi. Do not invent live prices.

**Demo (green box — run with me, ~5 min).** ETH-like dances harder than SPY-like.



```python
# =============================================================================
# DEMO — Beta is a dance partner, not a seatbelt (run with me)
# CONCEPT: An index is a "playlist." Beta ≈ how hard THIS name danced with the
#          playlist in THIS tiny sample. It is a risk-desk habit number —
#          not insurance and not a promise about tomorrow.
# =============================================================================

# Five handmade daily returns (fractions). Same length for asset and market.
# No wifi / live prices — if the network dies, we celebrate.
market = [0.02, -0.01, 0.03, 0.00, -0.02]   # toy index / "playlist"
ethish = [0.04, -0.02, 0.06, 0.01, -0.03]   # tends to dance harder
spyish = [0.01, 0.00, 0.01, 0.00, -0.01]    # calmer dancer


def beta(asset, market):
    """Sample beta ≈ cov(asset, market) / var(market) over this window.

    - Bigger |beta| → moved harder with the playlist in *this* sample.
    - Same formula a desk uses in spirit; our numbers are classroom toys.
    """
    n = len(asset)
    ma = sum(asset) / n                          # average asset return
    mm = sum(market) / n                         # average market return
    cov = sum((a - ma) * (m - mm) for a, m in zip(asset, market)) / (n - 1)
    var = sum((m - mm) ** 2 for m in market) / (n - 1)
    return cov / var


print("ETH-like beta:", round(beta(ethish, market), 2))
print("SPY-like beta:", round(beta(spyish, market), 2))
print("A beta is a habit in this sample, not a promise.")
```

**What you just saw.** Beta says how hard this coin danced with the playlist. It is not insurance.

### Keep this straight

**If you remember one thing.** Beta of ~2 means it *tended* to jump twice as hard in *this* tiny sample. Not a promise.

**Common mix-up.** “High beta means a good investment.” It means a wild dance partner, not a seatbelt.

**Why Eagecoin cares.** Before you pitch EAGE, ask *use* (will a classmate pay with it?), not “our beta beats Bitcoin.” Freeze ticker before Remix.

## 2. Same concert, three tickets

**Run with me.** Do not skip ahead to YOUR TURN yet.

**Real life.** “I have Bitcoin exposure” can mean three **different balance-sheet
objects** — an auditor grades them differently:

1. MetaMask → she holds the thing (**wallet / spot**).
2. Spot-Bitcoin ETF → she holds a *fund share* (custodian holds coins).
3. “0.1 BTC” in a trading app → she may only hold an **IOU** / customer claim.

If the app dies, (1) still works if she has the seed, (2) is a securities
claim, (3) is a customer-in-line story.

**Demo (green box — run with me, ~5 min).** A tiny classifier. Labels, not law.



```python
# =============================================================================
# DEMO — Same concert, three tickets (classroom labels only — not legal advice)
# CONCEPT: "I have Bitcoin exposure" can mean THREE different balance-sheet
#          objects. An auditor grades them differently:
#            wallet   = self-custody (you hold the thing)
#            etf      = fund share (a company holds seats for clients)
#            exchange = tab / IOU (the company may owe you the coin)
# =============================================================================


def package(kind: str) -> str:
    """Map a classroom label to a one-line custody / claim story."""
    kind = kind.lower()
    if kind == "wallet":
        return "self-custody -- you hold the thing"            # keys on your device
    if kind == "etf":
        return "fund share -- a company holds seats"           # you own a share, not the coin
    if kind == "exchange":
        return "tab / IOU -- the company may owe you the coin"  # customer claim
    return "unknown package"


for k in ["wallet", "etf", "exchange"]:
    print(f"{k:10} -> {package(k)}")
```

**What you just saw.** “I have Bitcoin exposure” is three different legal objects. Do not grade them as the same.

### Keep this straight

**If you remember one thing.** How you *get* the coin (airdrop, sale, earn) is a financial object, not a vibe.

**Common mix-up.** “Listed on an exchange” = “I hold the coin.” Often you hold a tab.

**Why Eagecoin cares.** Freeze ticker and distribution before Remix. Selling EAGE for CLASSUSD *plus* a 10× pitch is how you wander into Howey-shaped facts later.
