## 1. The serial number is the coin

Two teams can both name a token `EAGE`. Wallets care about the **contract
address**. Copy it from the class registry, not from memory.

**Demo.** Planned mint must fit the cap. Then every mailbox gets the same gift.



```python
# ===== Demo — yearbook stickers (cap math, then mailboxes) =====
# Every line has a plain-English comment — students are not Python experts.

ROSTER = ["Ava", "Ben", "Cara", "Dev"]  # who gets a gift (stand-in for classmate wallets)
AIRDROP = 200  # gift size each roster name gets at the start
POOL_RESERVE = 2000  # coins reserved for the trading bowl (TEAM / CLASSUSD)
TREASURY = 2000  # team stash kept to honor real-world promises
CAP = 10000  # box lid from A2 — never mint more than this

planned = AIRDROP * len(ROSTER) + POOL_RESERVE + TREASURY  # gifts + bowl + stash = left side of the inequality
print("planned", planned, "fits cap?", planned <= CAP)  # True only if the plan fits under the lid

balances = {n: 0 for n in ROSTER}  # start every mailbox at zero coins
for n in ROSTER:  # walk the roster one name at a time
    balances[n] += AIRDROP  # drop the same gift into this mailbox
print("mailboxes", balances)  # each name should show 200
print("ticker is a nickname; the registry address is the coin")  # remind: address = asset

```

**What you just saw.** The contract address is the coin. Cap math is airdrop × N + pool + treasury ≤ lid.
