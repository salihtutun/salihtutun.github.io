## 1. The serial number is the coin

Two teams can both name a token `EAGE`. Wallets care about the **contract
address**. Copy it from the class registry, not from memory.

**Demo.** Planned mint must fit the cap. Then every mailbox gets the same gift.



```python
# Demo — yearbook stickers. Planned mint vs box lid, then mailboxes.

ROSTER = ["Ava", "Ben", "Cara", "Dev"]
AIRDROP = 200
POOL_RESERVE = 2000
TREASURY = 2000
CAP = 10000

planned = AIRDROP * len(ROSTER) + POOL_RESERVE + TREASURY
print("planned", planned, "fits cap?", planned <= CAP)

balances = {n: 0 for n in ROSTER}
for n in ROSTER:
    balances[n] += AIRDROP
print("mailboxes", balances)
print("ticker is a nickname; the registry address is the coin")

```

**What you just saw.** The contract address is the coin. Cap math is airdrop × N + pool + treasury ≤ lid.

### Keep this straight

**If you remember one thing.** The contract address is the coin. The ticker is a nickname.
