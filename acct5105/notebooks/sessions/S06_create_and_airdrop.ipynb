## 1. The serial number is the coin

Two teams can both name a token `EAGE`. Wallets care about the **contract
address**. Copy it from the class registry, not from memory.

**Demo.** Planned mint must fit the cap. Then every mailbox gets the same gift.



```python
# Demo — yearbook stickers. Planned mint vs box lid, then mailboxes.

ROSTER = ["Ava", "Ben", "Cara", "Dev"]  # gift list size
AIRDROP = 200  # gift per roster address
POOL_RESERVE = 2000  # coins in the trading bowl
TREASURY = 2000  # team stash for promises
CAP = 10000  # lid on the ticket box

planned = AIRDROP * len(ROSTER) + POOL_RESERVE + TREASURY  # gift list size
print("planned", planned, "fits cap?", planned <= CAP)  # print for the PDF or screen

balances = {n: 0 for n in ROSTER}  # gift list size
for n in ROSTER:  # loop each item
    balances[n] += AIRDROP  # run this step
print("mailboxes", balances)  # print for the PDF or screen
print("ticker is a nickname; the registry address is the coin")  # print for the PDF or screen

```

**What you just saw.** The contract address is the coin. Cap math is airdrop × N + pool + treasury ≤ lid.
