## Words we use today

## 1. Coat vs stub, then both legs or neither

Team A: “1 EAGE *is* 5 minutes.” That is **the thing**.
Team B: “We bought 50 cookies. Each EAGE is a stub.” The cookies are the
coat. If the team eats the cookies, the stub is stationery.

Atomic swap: Ava has 6 EAGE, Ben has 3 CLASSUSD. Either **both** ledgers
update or *neither* does.

**Demo.** First swap succeeds. Second swap is broke — nobody moved.



```python
# Demo — atomic swap: both legs or NEITHER.

def atomic_swap(wallets, a, b, a_gives, a_amt, b_gives, b_amt):
    if wallets[(a, a_gives)] < a_amt or wallets[(b, b_gives)] < b_amt:
        print("SWAP REVERT -- nobody moved")
        return False
    wallets[(a, a_gives)] = wallets[(a, a_gives)] - a_amt
    wallets[(b, a_gives)] = wallets.get((b, a_gives), 0) + a_amt
    wallets[(b, b_gives)] = wallets[(b, b_gives)] - b_amt
    wallets[(a, b_gives)] = wallets.get((a, b_gives), 0) + b_amt
    print("SWAP OK")
    return True


w = {
    ("Ava", "EAGE"): 10,
    ("Ava", "CLASSUSD"): 0,
    ("Ben", "EAGE"): 0,
    ("Ben", "CLASSUSD"): 4,
}
atomic_swap(w, "Ava", "Ben", "EAGE", 6, "CLASSUSD", 3)
print("after OK", w)
atomic_swap(w, "Ava", "Ben", "EAGE", 6, "CLASSUSD", 99)
print("after revert", w)

```

**What you just saw.** Atomic means both legs move or neither. A chain does not put a coat in the closet.
