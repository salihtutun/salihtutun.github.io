## 1. Coat vs stub, then both legs or neither

Team A: “1 EAGE *is* 5 minutes.” That is **the thing**.
Team B: “We bought 50 cookies. Each EAGE is a stub.” The cookies are the
coat. If the team eats the cookies, the stub is stationery.

Atomic swap: Ava has 6 EAGE, Ben has 3 CLASSUSD. Either **both** ledgers
update or *neither* does.

**Demo.** First swap succeeds. Second swap is broke — nobody moved.



```python
# Demo — atomic swap: both legs or NEITHER.

def atomic_swap(wallets, a, b, a_gives, a_amt, b_gives, b_amt):  # define helper atomic_swap
    if wallets[(a, a_gives)] < a_amt or wallets[(b, b_gives)] < b_amt:  # branch when this is true
        print("SWAP REVERT -- nobody moved")  # print for the PDF or screen
        return False  # hand back the result
    wallets[(a, a_gives)] = wallets[(a, a_gives)] - a_amt  # run this step
    wallets[(b, a_gives)] = wallets.get((b, a_gives), 0) + a_amt  # run this step
    wallets[(b, b_gives)] = wallets[(b, b_gives)] - b_amt  # run this step
    wallets[(a, b_gives)] = wallets.get((a, b_gives), 0) + b_amt  # run this step
    print("SWAP OK")  # print for the PDF or screen
    return True  # hand back the result


w = {  # set w
    ("Ava", "EAGE"): 10,  # run this step
    ("Ava", "CLASSUSD"): 0,  # run this step
    ("Ben", "EAGE"): 0,  # run this step
    ("Ben", "CLASSUSD"): 4,  # run this step
}  # close the bracket above
atomic_swap(w, "Ava", "Ben", "EAGE", 6, "CLASSUSD", 3)  # call the helper
print("after OK", w)  # print for the PDF or screen
atomic_swap(w, "Ava", "Ben", "EAGE", 6, "CLASSUSD", 99)  # call the helper
print("after revert", w)  # print for the PDF or screen

```

**What you just saw.** Atomic means both legs move or neither. A chain does not put a coat in the closet.
