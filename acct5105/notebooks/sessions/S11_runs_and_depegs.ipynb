A **peg** says "this chip should stay worth $1."
A **depeg** is the moment the market laughs.
A **run** is everyone hitting redeem at once — a fire drill with one door.



## Words we use today

Read this once. If a later cell uses the word, come back here.

| Word | In this class it means |
|---|---|
| **Peg** | A promise that the chip stays worth a number (often $1, here maybe 0.50 CLASSUSD). |
| **Depeg** | The market price walks away from that number. |
| **Run** | Everyone redeems at once. One door, Saturday rush. |
| **Depth** | How much CLASSUSD is actually in the bowl. Thin bowl → wild quotes. |
| **Contagion** | Your pause or depeg bricks *other* coins that priced through you. |
| **Kill criteria** | If-then with numbers. Not “we’ll watch the market.” |



## 1. Saturday rush

10,000 gift cards, 200 scoops. If everyone ran *tonight*, each card is
really 200/10,000 = **0.02 scoops**. The poster still says “1 card = 1 scoop.”

**Demo.** Honor who shows up until the freezer is empty.



```python
# Demo -- gift-card rush. The poster is the peg. The 0.02 is the run price.

scoops = 200
cards = 10000
print("if everyone ran, each card is really", scoops / cards, "scoops")


def rush(shown, freezer, outstanding):
    honored = min(shown, freezer)
    freezer = freezer - honored
    outstanding = outstanding - honored
    print("shown", shown, "honored", honored, "freezer", freezer, "cards left", outstanding)
    return freezer, outstanding


scoops, cards = rush(150, scoops, cards)
scoops, cards = rush(80, scoops, cards)
print("Pause does not refill the freezer.")

```

**What you just saw.** A peg is a sentence. A run is a crowd. Pause stops transfers; it does not restock scoops.
