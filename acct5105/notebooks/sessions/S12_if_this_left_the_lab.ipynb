## 1. Two posters

Poster A: “1 EAGE = one cookie at office hours. Not for sale to strangers.”
Poster B: “Buy EAGE tonight — we will list it and it 10xes.”

Classroom scoring (not law): public sale +2, profit pitch +3, team-works-price +2.
Score ≥ 5 → cut the pitch before you write A6.

**Demo.** Score both posters.



```python
# Demo -- classroom labels, not legal advice.

def howey_score(facts: dict) -> int:  # define helper howey_score
    score = 0  # set score
    if facts["sold_to_public"]:  # branch when this is true
        score += 2  # bump the counter
    if facts["profit_pitch"]:  # branch when this is true
        score += 3  # bump the counter
    if facts["team_will_work_on_price"]:  # branch when this is true
        score += 2  # bump the counter
    return score  # hand back the result


poster_a = {  # set poster a
    "sold_to_public": False,  # run this step
    "profit_pitch": False,  # run this step
    "team_will_work_on_price": False,  # run this step
    "redeem_for_a_real_good": True,  # run this step
}  # close the bracket above
poster_b = {  # set poster b
    "sold_to_public": True,  # run this step
    "profit_pitch": True,  # run this step
    "team_will_work_on_price": True,  # run this step
    "redeem_for_a_real_good": True,  # run this step
}  # close the bracket above

for name, facts in [("cookie voucher", poster_a), ("10x pitch", poster_b)]:  # loop each item
    s = howey_score(facts)  # set s
    vibe = "MIXED / security-like -- cut the pitch" if s >= 5 else "closed voucher -- still no mainnet"  # set vibe
    print(name, "score", s, "->", vibe)  # print for the PDF or screen

```

**What you just saw.** Facts classify the object. A ticker and a tweet do not. Still testnet.
