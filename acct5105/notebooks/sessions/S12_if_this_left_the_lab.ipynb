## 1. Two posters

Poster A: “1 EAGE = one cookie at office hours. Not for sale to strangers.”
Poster B: “Buy EAGE tonight — we will list it and it 10xes.”

Classroom scoring (not law): public sale +2, profit pitch +3, team-works-price +2.
Score ≥ 5 → cut the pitch before you write A6.

**Demo.** Score both posters.



```python
# Demo -- classroom labels, not legal advice.

def howey_score(facts: dict) -> int:
    score = 0
    if facts["sold_to_public"]:
        score += 2
    if facts["profit_pitch"]:
        score += 3
    if facts["team_will_work_on_price"]:
        score += 2
    return score


poster_a = {
    "sold_to_public": False,
    "profit_pitch": False,
    "team_will_work_on_price": False,
    "redeem_for_a_real_good": True,
}
poster_b = {
    "sold_to_public": True,
    "profit_pitch": True,
    "team_will_work_on_price": True,
    "redeem_for_a_real_good": True,
}

for name, facts in [("cookie voucher", poster_a), ("10x pitch", poster_b)]:
    s = howey_score(facts)
    vibe = "MIXED / security-like -- cut the pitch" if s >= 5 else "closed voucher -- still no mainnet"
    print(name, "score", s, "->", vibe)

```

**What you just saw.** Facts classify the object. A ticker and a tweet do not. Still testnet.
